```python
crime_rent_data['high_rent_growth'] = crime_rent_data['rent_change'].apply(lambda x: 'yes' if x > 0.334 else 'no')
```

```python
crime_rent_data['is_gentrified'] = np.where(
    crime_rent_data['high_rent_growth'] == 'yes',
    'yes',
    'no'
)
```

```python
train = crime_rent_data.loc[crime_rent_data['year'].isin([2019,2020,2021,2022])].set_index('Incident Zip')
test = crime_rent_data.loc[crime_rent_data['year']==2023].set_index('Incident Zip')

training_feats =  [
'After Hours - Licensed Est', 
'Blocked Bike Lane', 'Blocked Sidewalk', 'Broken Sidewalk', 
'Business not in compliance', 'Car/Truck Horn', 'Car/Truck Music','Chronic Stoplight Violation',
'Commercial Overnight Parking', 'Congestion/Gridlock', 'Disruptive Passenger',
'Drag Racing', 'Grass/Weeds', 'Homeless Issue', 'Hours of Operation', 'In Car',
'In Public', 'Lighting', 'Loud Music/Party', 'Loud Talking',
'Loud Television', 'Mass Gathering', 'Nuisance/Truant', 
'People Created Noise', 'Performance', 
'Sidewalk Violation', 'Street Con Game', 'Structure - Outdoors',
'Ticket Scalping', 'Tortured', 'Trespassing', 'Truck Route Violation',
]

prediction_var = 'is_gentrified'

Xtrain,ytrain = train[training_feats].values, train[prediction_var]
Xtest,ytest = test[training_feats].values, test[prediction_var]

# Convert 'no' to 0 and 'yes' to 1
ytrain = ytrain.map({'no': 0, 'yes': 1})
ytest = ytest.map({'no': 0, 'yes': 1})
```

```python
#make a new column in df_merged merging Year and Incident Zip with a _
crime_rent_data['merge_key'] = crime_rent_data['year'].astype(str) + '_' + crime_rent_data['Incident Zip'].astype(str)
#only include training features along with year, incident zip and is_gentrified for all years except 2011
crime_rent_data = crime_rent_data[crime_rent_data['year'] != 2011]
df_final = crime_rent_data[['year', 'Incident Zip', 'merge_key', 'is_gentrified'] + training_feats]
df_final
```

```python
from sklearn.ensemble import RandomForestRegressor,RandomForestClassifier
from sklearn.tree import DecisionTreeRegressor,DecisionTreeClassifier
from xgboost import XGBRegressor,XGBClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score,f1_score
import matplotlib.pyplot as plt
import numpy as np

def run_prediction(
        model, 
        train, 
        test,
        training_feats,
        prediction_var,
        params_config, 
        nm, 
        plot=False
    ):
    '''
    Performs a Grid Search on a set of parameters, finds an optimal model, and makes probabilistic predictions. 
    Assumes a binary classifier. 

    Inputs: 
        **model: a sklearn or XGBoost model object 
        **train: the training dataset 
        **test: the test dataset to make predictions on 
        **training_feats: the features you want to train the classifier on 
        **prediction_var: the variable we want to predict. should be boolean in nature 
        **params_config: a dictionary of hyperparameters to test the model 
        **nm: the name of the model 

    Outputs: A tuple consisting of the following: 
        ** the probabilities of class assignment (p(Y=0) and p(Y=1)) per item in the test set 
        ** the feature importance scores associated with each trained model 
        ** the best model after the gridsearch 
    
    
    '''

    #build the training/prediction data
    Xtrain,ytrain = train[training_feats].values, train[prediction_var]
    Xtest,ytest = test[training_feats].values, test[prediction_var]

    #grid search CV
    model_cv = GridSearchCV(estimator=model, param_grid=params_config, cv=3)

    #fits/predicts the optimized model
    model_cv.fit(Xtrain, ytrain)
    
    predictions = model_cv.predict(Xtest)
    train['predictions'] = model_cv.predict_proba(Xtrain)[:,1]
    test['predictions'] = model_cv.predict_proba(Xtest)[:,1]

    corr = np.corrcoef(test['rent_change'], test['predictions'])[0][1]
    if plot: 
        _, ax = plt.subplots()
        ax.scatter(
            test.loc[test[prediction_var]]['predictions'], 
            test.loc[test[prediction_var]]['rent_change'],
            label='Gentrified'
        )
        ax.scatter(
            test.loc[~test[prediction_var]]['predictions'], 
            test.loc[~test[prediction_var]]['rent_change'],
            label='Not Gentrified'
        )
        ax.axvline(0.5,label='Classifier Decision Boundary',color='red')
        ax.set_title(f'{nm}: Corr[Gentrification Score, Rent Change]={corr}')
        ax.set_xlabel('Probability (Gentrification Hotspot Score)')
        ax.set_ylabel('Rent Pct Change')
        ax.legend()
        plt.show()

    #print out various metrics (accuracy + F1 -- weighted prec. and recall)
    print(f'*** {nm} Accuracy: {accuracy_score(ytest, predictions)} ***')
    print(f'*** {nm} F1: {f1_score(ytest, predictions)} ***')
    
    if nm == '[BASELINE] NaiveBayes':
        return train, test, {}, model_cv.best_estimator_, corr

    feature_importances = model_cv.best_estimator_.feature_importances_
    idx_sort_topfeats = np.argsort(feature_importances)[::-1][:20]
    feat_importance_scores = {training_feats[i]:feature_importances[i] for i in idx_sort_topfeats}
    
    return train, test, feat_importance_scores, model_cv.best_estimator_, corr
    

 
```

```python
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from xgboost import XGBRegressor, XGBClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import Lasso
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

def run_prediction(model, params_config, nm, Xtrain, Xtest, ytrain, ytest, training_feats):
    model_cv = GridSearchCV(estimator=model, param_grid=params_config, cv=3)
    model_cv.fit(Xtrain, ytrain)
    predictions = model_cv.predict(Xtest)
    
    try:
        probs = model_cv.predict_proba(Xtest)
    except AttributeError:
        probs = None

    print(f'*** {nm} Accuracy: {accuracy_score(ytest, predictions)} ***')
    print(f'*** {nm} F1: {f1_score(ytest, predictions)} ***')

    if nm == '[BASELINE] NaiveBayes':
        return probs, {}, None

    feature_importances = model_cv.best_estimator_.feature_importances_
    idx_sort_topfeats = np.argsort(feature_importances)[::-1][:20]
    feat_importance_scores = {training_feats[i]: feature_importances[i] for i in idx_sort_topfeats}

    return probs, feat_importance_scores, model_cv.best_estimator_


models = [
    {
        'model': GaussianNB(),  # no balancing needed
        'params_config': {},
        'nm': '[BASELINE] NaiveBayes'
    },
    {
        'model': DecisionTreeClassifier(class_weight='balanced'),
        'params_config': {
            'max_depth': [3, 5, 7],
            'min_samples_split': [2, 5, 10],
            'min_samples_leaf': [1, 2, 5]
        },
        'nm': 'Single Decision Tree'
    },
    {
        'model': RandomForestClassifier(class_weight='balanced'),
        'params_config': {
            'n_estimators': [50, 100, 200, 300],
            'max_depth': [3, 5, 7],
            'min_samples_split': [2, 5, 10],
            'min_samples_leaf': [1, 2, 5]
        },
        'nm': 'Random Forests'
    },
    {
        'model': XGBClassifier(scale_pos_weight=1),  # later tuning scale_pos_weight better
        'params_config': {
            'n_estimators': [50, 100, 200, 300],
            'max_depth': [3, 5, 7, 9],
            'learning_rate': [0.01, 0.05, 0.1, 0.2]
        },
        'nm': 'XGBoost'
    }
]


for m_config in models:

    p, fi, m = run_prediction(
        model=m_config['model'],
        params_config=m_config['params_config'],
        nm=m_config['nm'],
        Xtrain=Xtrain, Xtest=Xtest,
        ytrain=ytrain, ytest=ytest,
        training_feats=training_feats
    )

    print(f'*** Feature Importance Scores *** \n {fi}')

    m_config['feature_importances'] = fi
    m_config['probs'] = p
    m_config['model'] = m
```

```python
import pandas as pd

results = []

for m_config in models:
    model_name = m_config['nm']

    probs = m_config['probs']

    predictions = (probs[:,1] > 0.5).astype(int) if probs.shape[1] == 2 else probs.argmax(axis=1)
    acc = accuracy_score(ytest, predictions)
    f1 = f1_score(ytest, predictions)

    # Feature importances (if available)
    top_feats = ', '.join(list(m_config['feature_importances'].keys())[:3]) if m_config['feature_importances'] else 'N/A'

    # Save everything in a dictionary
    results.append({
        'Model': model_name,
        'Accuracy': acc,
        'F1 Score': f1,
        'Top Features': top_feats
    })

# Create a nice summary DataFrame
results_df = pd.DataFrame(results)

# Sort by best F1 score (or Accuracy if you prefer)
results_df = results_df.sort_values('F1 Score', ascending=False).reset_index(drop=True)

# Show it
display(results_df)
```

```python
import geopandas as gpd
import pandas as pd
import plotly.express as px
import json
import numpy as np
from shapely.wkt import loads

geo_zip_nyc = pd.read_csv('Modified_Zip_Code_Tabulation_Areas.csv')
geo_zip_nyc['zip'] = geo_zip_nyc['MODZCTA'].astype(str)
geo_zip_nyc = geo_zip_nyc.rename(columns={'the_geom': 'geometry'})
geo_zip_nyc['geometry'] = geo_zip_nyc['geometry'].apply(loads)

geo_zip_nyc = gpd.GeoDataFrame(geo_zip_nyc, geometry='geometry')
geo_zip_nyc.set_index('zip', inplace=True)
geo_zip_nyc.set_crs(epsg=4326, inplace=True)

best_model_result = models[0]  

# Step 3: Prepare hotspot scores
test.index = test.index.astype(str)  
hotspot_scores = pd.DataFrame(
    best_model_result['probs'][:, 1],
    index=test.index,
    columns=['Gentrification Hotspot Score']
)

w_geo = gpd.GeoDataFrame(
    pd.concat([geo_zip_nyc, hotspot_scores], axis=1).dropna(),
    crs="EPSG:4326"
)

w_geo['zip'] = w_geo.index  # add zip column
w_geo = w_geo.reset_index(drop=True)  # reset index but keep zip

geojson = json.loads(w_geo.to_json())

w_geo_projected = w_geo.to_crs(epsg=3857)
center_lat = w_geo_projected.centroid.to_crs(epsg=4326).y.mean()
center_lon = w_geo_projected.centroid.to_crs(epsg=4326).x.mean()

fig = px.choropleth_mapbox(
    w_geo,
    geojson=geojson,
    locations="zip",  
    color="Gentrification Hotspot Score",
    featureidkey="properties.zip",
    center={"lat": center_lat, "lon": center_lon},
    zoom=10,
    mapbox_style="carto-positron",
    color_continuous_scale="plasma", 
    hover_name="zip",
    hover_data={"Gentrification Hotspot Score": True}
)

fig.update_layout(
    coloraxis_colorbar=dict(
        title="Gentrification Hotspot Score",
        tickformat=".2f"
    ),
    coloraxis_cmin=0.2, 
    coloraxis_cmax=1.0,  
    margin={"r": 0, "t": 0, "l": 0, "b": 0}
)

fig.write_html("gentrification_hotspot_map.html")



```
