```python
def process_income_response(response, year):
    """
    Convert Census API response to a DataFrame with selected columns and a year column.

    Parameters:
        response: requests.Response object from Census API
        year (str or int): Year of the data

    Returns:
        pd.DataFrame: Cleaned DataFrame with ['ucgid', 'year', 'S1901_C01_012E', 'S1901_C01_013E']
    """
    df = pd.DataFrame(response.json()[1:], columns=response.json()[0])
    df = df[['ucgid', 'S1901_C01_001E','S1901_C01_012E', 'S1901_C01_013E']]
    df['year'] = str(year)
    return df[['ucgid', 'year', 'S1901_C01_001E','S1901_C01_012E', 'S1901_C01_013E']]
```

```python
def process_rent2_response(response, year):
    """
    Convert Census API response to a DataFrame with selected columns and a year column.

    Parameters:
        response: requests.Response object from Census API
        year (str or int): Year of the data

    Returns:
        pd.DataFrame: Cleaned DataFrame with ['ucgid', 'year', 'S1901_C01_012E', 'S1901_C01_013E']
    """
    df = pd.DataFrame(response.json()[1:], columns=response.json()[0])
    df = df[['ucgid', 'DP04_0124E','DP04_0132E','DP04_0079E','DP04_0088E']]
    df['year'] = str(year)
    return df[['ucgid', 'year', 'DP04_0124E','DP04_0132E', 'DP04_0079E','DP04_0088E']]
```
